```python
from nltk.corpus import brown, stopwords
from nltk.tokenize import sent_tokenize, word_tokenize
```

```python
brown_sentences = brown.sents()
selected_sentences = [" ".join(sentence) for sentence in brown_sentences[:20]]
print("Number of sentences selected:", len(selected_sentences))
```

```python
text = " ".join(selected_sentences)
```

```python
sentence_tokens = sent_tokenize(text)
for i, sentence in enumerate(sentence_tokens[:5], 1):
    print(f"{i}. {sentence}")
```

```python
word_tokens = word_tokenize(text)
print(word_tokens[:50])
```

```python
lower_tokens = [word.lower() for word in word_tokens]
alphabetic_tokens = [
    word for word in lower_tokens
    if word.isalpha()
]
```

```python
stop_words = set(stopwords.words('english'))

cleaned_tokens = [
    word for word in alphabetic_tokens
    if word not in stop_words
]
print(cleaned_tokens[:50])
```

```python
selected_words = alphabetic_tokens[:30]
print(selected_words)
```

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
stemmed_words = [
    stemmer.stem(word)
    for word in selected_words
]
for original, stemmed in zip(selected_words, stemmed_words):
    print(original, "→", stemmed)
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in selected_words
]
for original, lemma in zip(selected_words, lemmatized_words):
    print(original, "→", lemma)
```

```python
required_words = [
    "playing",
    "studies",
    "running",
    "better",
    "cars"
]

words_for_analysis = selected_words + required_words
print("Words for Stemming and Lemmatization:")
print(words_for_analysis)
```

```python
lemmatized_words = []

for word in words_for_analysis:

    # Give correct POS for required words
    if word == "playing":
        lemma = lemmatizer.lemmatize(word, pos="v")

    elif word == "studies":
        lemma = lemmatizer.lemmatize(word, pos="v")

    elif word == "running":
        lemma = lemmatizer.lemmatize(word, pos="v")

    elif word == "better":
        lemma = lemmatizer.lemmatize(word, pos="a")

    elif word == "cars":
        lemma = lemmatizer.lemmatize(word, pos="n")

    else:
        lemma = lemmatizer.lemmatize(word)

    lemmatized_words.append(lemma)
```

```python
from nltk.corpus import words
word_list = words.words()
print(word_list[:20])
print("\nTotal words:", len(word_list))
```

```python
misspelled_words = [
    "recieve",
    "definately",
    "seperate",
    "occured",
    "enviroment",
    "beleive",
    "adress",
    "goverment",
    "tommorow",
    "teh"
]

print("Misspelled Words:")

for word in misspelled_words:
    print(word)
```

```python
from difflib import get_close_matches

vocabulary = set(
    word.lower()
    for word in word_list
    if word.isalpha()
)

for word in misspelled_words:

    suggestions = get_close_matches(
        word,
        vocabulary,
        n=5,
        cutoff=0.5
    )

    print(f"{word} -> {suggestions}")
```

```python
def correct_sentence(sentence):
    
    tokens = word_tokenize(sentence)
    corrected_tokens = []
    
    for word in tokens:
        
        if word.isalpha():
            
            lower_word = word.lower()
            
            if lower_word not in vocabulary:
                
                suggestions = get_close_matches(
                    lower_word,
                    vocabulary,
                    n=1,
                    cutoff=0.6
                )
                
                if suggestions:
                    corrected_tokens.append(suggestions[0])
                else:
                    corrected_tokens.append(word)
            else:
                corrected_tokens.append(word)
        
        else:
            corrected_tokens.append(word)
    
    return " ".join(corrected_tokens)
```

```python
sentences = [
    "I definately want to visit the museum.",
    "Please recieve the document tommorow.",
    "The goverment should protect the enviroment.",
    "I beleive this is the correct adress.",
    "The meeting occured in a seperate room."
]



for word in misspelled_words:
    
    suggestions = get_close_matches(
        word,
        vocabulary,
        n=1,
        cutoff=0.6
    )
    
    if suggestions:
        print(f"{word:15} -> {suggestions[0]}")
    else:
        print(f"{word:15} -> No suggestion found")




for i, sentence in enumerate(sentences, 1):
    
    corrected = correct_sentence(sentence)
    
    print(f"\nSentence {i}")
    print("Original :", sentence)
    print("Corrected:", corrected)
```

```python
brown_words = brown.words()

brown_subset = brown_words[:10000]

print(len(brown_subset))
```

```python
tokens = [
    word.lower()
    for word in brown_subset
    if word.isalpha()
]

print(len(tokens))
print(tokens[:50])
```

```python
from nltk.util import ngrams
from nltk.probability import FreqDist
unigrams = list(ngrams(tokens, 1))

bigrams = list(ngrams(tokens, 2))

trigrams = list(ngrams(tokens, 3))


print(unigrams[:10])

print(bigrams[:10])

print(trigrams[:10])
```

```python

unigram_freq = FreqDist(unigrams)

bigram_freq = FreqDist(bigrams)

trigram_freq = FreqDist(trigrams)
```

```python
top_unigrams = unigram_freq.most_common(10)
unigram_df = pd.DataFrame(
    top_unigrams,
    columns=["Unigram", "Frequency"]
)

top_bigrams = bigram_freq.most_common(10)
bigram_df = pd.DataFrame(
    top_bigrams,
    columns=["Bigram", "Frequency"]
)

top_trigrams = trigram_freq.most_common(10)
trigram_df = pd.DataFrame(
    top_trigrams,
    columns=["Trigram", "Frequency"]
)
```

```python
print(unigram_df)
print(bigram_df)
print(trigram_df)
```
